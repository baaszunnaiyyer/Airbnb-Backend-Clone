# **Imputation Performance**

##### performing imputation operations to fill missing vlaues with meaning full Values Ensuring that the dataset remains usable for analysis and machine learning models.

### **Fill Missing Values with a Constant (Fixed Value)**

````python
from sklearn.impute import SimpleImputer

# Replace NaN with a fixed value (e.g., 0)
imputer = SimpleImputer(strategy='constant', fill_value=0)
train_df['A'] = imputer.fit_transform(train_df[['A']])
test_df['A'] = imputer.transform(test_df[['A']])

print(train_df)
````

### **Fill Missing Values with Mean, Median, or Mode**

````python
imputer = SimpleImputer(strategy='mean')  # Can be 'median' or 'most_frequent' (mode)
train_df['A'] = imputer.fit_transform(train_df[['A']])
test_df['A'] = imputer.transform(test_df[['A']])

print(train_df)
````

### **K-Nearest Neighbors (KNN) Imputation**

Uses similar data points (neighbors) to predict the missing values.

````python
from sklearn.impute import KNNImputer

imputer = KNNImputer(n_neighbors=3)  # Uses 3 nearest neighbors
train_df_filled = imputer.fit_transform(train_df)
test_df_filled = imputer.transform(test_df)

# Convert back to DataFrame with column names
train_df_filled = pd.DataFrame(train_df_filled, columns=train_df.columns)
test_df_filled = pd.DataFrame(test_df_filled, columns=test_df.columns)

print(train_df_filled)
````

## **Encoding**

#### **One Hot Encoding**

##### **Way One**

````python
train_df = pd.get_dummies(train_df, columns=['category_column'], drop_first=True)
test_df = pd.get_dummies(test_df, columns=['category_column'], drop_first=True)
````

##### **Second Way**

````python
from sklearn.preprocessing import OneHotEncoder
from sklear.compose import ColumnTransformer

ct = ColumnTransformer([('encoder', OneHotEncoder(), ['category_column'])], remainder='passthrough')
train_df = ct.fit_transform(train_df)
test_df = ct.transform(test_df)
````

#### **Binary Encoding**

##### **Way One**

````python
train_df['category_column'].replace({'yes': 1, 'no': 0}, inplace=True)
test_df['category_column'].replace({'yes': 1, 'no': 0}, inplace=True)
````

##### **Second Way**

````python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
train_df['category_column'] = le.fit_transform(train_df['category_column'])
test_df['category_column'] = le.transform(test_df['category_column'])
````
